## Introdução

### Contextualização

Um dos maiores desafios para engenheiros nos tempos atuais é desenvolver técnicas capazes de estimar o estado de um sistema dinâmico quando o mesmo está repleto de incertezas.\
Por isso, métodos que mesclam a certeza das equações matemáticas e a capacidade de detectar perturbações dos sensores e minimizar as fraquezas dos sensores que é sua suscetibilidade a ruídos e limitações de resolução e precisão são muito valorizados na engenharia.\
Algumas das aplicações que levou ao aumento da notoriedade deste tipo de filtro é a navegação aeronáutica, sistemas de rastreamento de alvos de radares, navegação e acoplamento de naves espaciais.

### Histórico

O filtro de Kalman foi desnvolvido pelo matemático húngaro-americano Rudolf Emil Kalman nos anos 60 e foi publicado em um artigo chamado: _A New Approach to Linear Filtering and Prediction Problems_.\
Um diferencial dos demais métodos da época, o novo algorítimo não precisava processar todo o histórico para fazer uma estimativa, ele conseguia atualizar sua estimativa apenas usando a medição atual e a anterior, assim reduzindo o custo computacional do programa.


### Justificativa

A justificativa para o estudo do Filtro de Kalman está diretamente relacionada ao impacto que proporciona em sistemas de engenharia. Em processos industriais, por exemplo, sensores de temperatura instalados em fornos, caldeiras e reatores frequentemente apresentam oscilações decorrentes de ruídos elétricos e perturbações ambientais. A utilização do filtro permite reduzir essas flutuações sem introduzir atrasos significativos, fornecendo estimativas mais suaves e confiáveis para os controladores.

### Estado da arte

Aqui estão alguns desenvolvimentos que se tiveram com a popularização das inteligências artificiaise e avanços na robótica, automóveis autônomos e a indústria 4.0:

Segundo pesquisa publicada pelo grupo _AI-Aided Kalman Filters_, disponibilizada na plataforma arXiv em outubro de 2024, uma das maiores tendências da área é a utilização de redes neurais para estimar automaticamente parâmetros do Filtro de Kalman, como as matrizes de covariância dos ruídos de processo e de medição. Os pesquisadores destacam que essa abordagem híbrida permite combinar o conhecimento físico do sistema com técnicas modernas de aprendizado profundo, melhorando o desempenho em sistemas altamente não lineares e sujeitos a mudanças de comportamento (SHLEZINGER et al., 2024).

De acordo com um estudo publicado pela revista científica _Measurement_, em setembro de 2024, pesquisadores desenvolveram um Filtro de Kalman adaptativo baseado em inferência bayesiana e no algoritmo Expectation-Maximization. Os resultados mostraram que o método foi capaz de reduzir significativamente os erros de posicionamento em sistemas de navegação utilizando sensores MEMS de baixo custo, alcançando erros inferiores a 2,8 metros mesmo em ambientes urbanos complexos (CHEN; LI; DU, 2024).

Uma curiosidade interessante é que, apesar de ter sido desenvolvido há mais de sessenta anos, o Filtro de Kalman continua sendo amplamente utilizado nas tecnologias mais modernas. Segundo pesquisadores do _Google DeepMind_, em trabalho publicado durante a ICLR 2024, conceitos derivados das equações do Filtro de Kalman foram empregados em algoritmos de aprendizado contínuo para redes neurais profundas. O objetivo foi permitir que os modelos de Inteligência Artificial atualizassem seus parâmetros continuamente à medida que novos dados fossem recebidos, demonstrando que uma técnica desenvolvida em 1960 permanece relevante para pesquisas atuais em IA (GALASHOV et al., 2024).

## Fundamentação Teórica

### Estimação de dados

Segundo Dan Simon (2006), a estimação de estados consiste em determinar, da forma mais precisa possível, as variáveis internas de um sistema utilizando simultaneamente um modelo matemático e as medições disponíveis.Esse processo permite reduzir a influência dos ruídos presentes nos sensores, aumentando a confiabilidade das informações utilizadas pelos sistemas de controle.\
O conceito de estado representa o menor conjunto de variáveis capaz de descrever completamente a evolução dinâmica de um sistema em um determinado instante. Em um motor elétrico, por exemplo, velocidade angular e corrente podem representar os estados do sistema. Em um forno industrial, temperatura interna, fluxo de calor e taxa de aquecimento podem constituir as variáveis de estado.

### Sistemas dinâmicos

Um sistema dinâmico é aquele que varia seu comportamento ao longo do tempo dependendo de suas condições iniciais, entradas externas e características físicas. Esses sistemas podem ser representados por equações diferenciais no tempo contínuo ou por equações de diferenças no tempo discreto.\
Na forma discreta, utilizada pelo Filtro de Kalman, um sistema linear é representado por:

$x_k=A x_{k-1}+B u_k+w_k$

onde:

($x_k$) representa o vetor de estados;  
($A$) é a matriz de transição de estados;  
($B$) representa a influência das entradas;  
($u_k$) corresponde às entradas do sistema;  
($w_k$) representa o ruído de processo.  

E as medições realizadas pelos sensores são descritas por:

$z_k=H x_k+v_k$

em que:

($z_k$) é o vetor de medições;  
($H$) é a matriz de observação;  
($v_k$) representa o ruído de medição.  

Esses dois modelos constituem a base matemática do Filtro de Kalman.

## Ruídos em sistemas de medição

Todo sistema de medição está sujeito a algum tipo de ruído, na teoria do Filtro de Kalman são considerados dois tipos principais de ruído:

__Ruído de Processo ($w_k$)__

Representa as incertezas do próprio modelo matemático. Mesmo conhecendo a dinâmica do sistema, sempre existem fenômenos não modelados ou perturbações que afetam sua evolução.

__Ruído de Medição ($v_k$)__

Está associado às imperfeições dos sensores e dos equipamentos de aquisição de dados.


Admite-se que ambos possuem distribuição Gaussiana de média nula:

$w_k \sim N(0,Q)$

$v_k \sim N(0,R)$

onde:

($Q$) é a matriz de covariância do ruído de processo;  
($R$) é a matriz de covariância do ruído de medição.  

Essas matrizes são responsáveis por determinar o quanto o filtro confia no modelo matemático ou nas medições realizadas.

### Conceitos do Filtro de Kalman 

#### Predição

Inicialmente o algoritmo utiliza o modelo matemático para prever o estado futuro do sistema.

As equações são:


$\hat{x}k^- = A\hat{x}{k-1}+Bu_k$

$P_k^- = AP_{k-1}A^T+Q$


onde:

($\hat{x}_k^-$) é o estado previsto;  
($P_k^-$) é a covariância prevista do erro.


#### Correção

Quando uma nova medição é recebida, a estimativa é atualizada.

Primeiramente calcula-se o ganho de Kalman:

$K_k=P_k^-H^T(H P_k^- H^T+R)^{-1}$

Em seguida, corrige-se o estado:

$\hat{x}_k=\hat{x}_k^-+K_k(z_k-H\hat{x}_k^-)$

Finalmente atualiza-se a covariância:

$P_k=(I-K_kH)P_k^-$

O ganho de Kalman determina o equilíbrio entre confiar na previsão do modelo ou na medição realizada.

#### Ganho de Kalman

O ganho de Kalman é considerado o elemento mais importante do algoritmo.

Quando o ruído dos sensores é elevado (valor elevado de (R)), o ganho diminui, fazendo com que o filtro confie mais no modelo matemático.

Por outro lado, quando o modelo apresenta grandes incertezas (valor elevado de (Q)), o ganho aumenta e o algoritmo passa a confiar mais nas medições.

Esse comportamento adaptativo é uma das principais vantagens do método em relação aos filtros convencionais.

#### Aplicações Industriais


O Filtro de Kalman possui ampla utilização na indústria devido à sua capacidade de fornecer estimativas confiáveis em tempo real.

Entre suas principais aplicações destacam-se:

* monitoramento de temperatura em fornos industriais;  
* estimação de velocidade em motores elétricos;  
* controle de robôs industriais;  
* fusão de sensores em veículos autônomos;  
* navegação de drones;  
* monitoramento de processos químicos;  
* sistemas de posicionamento por GPS;  
* previsão de trajetória de aeronaves;  

## Desenvolvimento da Aplicação

Com o objetivo de demonstrar a aplicação prática do Filtro de Kalman em sistemas industriais, foi desenvolvido um modelo computacional que simula o comportamento térmico de um forno industrial. A implementação foi realizada na linguagem Python, permitindo reproduzir tanto a dinâmica do processo quanto as incertezas normalmente encontradas em medições reais de temperatura.

A metodologia adotada consiste em comparar a temperatura real do forno, obtida a partir de um modelo matemático, com a temperatura medida por um sensor sujeito a ruídos. Em seguida, aplica-se o Filtro de Kalman para estimar a temperatura real do sistema, avaliando sua capacidade de reduzir oscilações nas medições e fornecer uma estimativa mais confiável para aplicações de controle.

O desenvolvimento foi dividido nas seguintes etapas:

* Modelagem matemática do forno industrial;
* Definição dos parâmetros físicos do processo;
* Simulação da evolução da temperatura do forno;
* Inserção de ruídos nas medições;
* Implementação do algoritmo do Filtro de Kalman;
* Comparação entre a temperatura real, a temperatura medida e a temperatura estimada;
* Avaliação do desempenho do filtro.
