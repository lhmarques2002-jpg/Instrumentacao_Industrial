## Definição Conceitual:

### "A vazão é uma grandeza física utilizada para medir a quantidade de fluido que passa por uma determinada seção de um sistema em um intervalo de tempo."
### Essa quantidade podendo ser descrita tanto pelo volume quanto pela massa do fluido, que por sua vez pode ser tanto um líquido quanto um gaz.
### A vazão volumétrica também pode ser obtida pelo resultado do produto da área seccional pela média da velocidade do fluido.

## Definição Matemática:
### para vazão volumétrica: <span style="font-size:30px">$Q_v = \frac{V}{t}$</span>  ou   <span style="font-size:30px">$  Q_v = S \cdot v$</span>

#### Onde:
<span style="font-size:21px">$Q_v$ = Vazão volumétrica</span>\
<span style="font-size:21px">V = volume</span>\
<span style="font-size:21px">t = tempo</span>\
<span style="font-size:21px">S = área</span>\
<span style="font-size:21px">v = velocidade</span>




   

### para vazão mássica: <span style="font-size:30px">$Q_m = \frac{m}{t}$</span>  ou <span style="font-size:30px">$  Q_m = Q_v \cdot \rho$</span>
#### Onde:
<span style="font-size:21px">$Q_m$ = Vazão mássica</span>\
<span style="font-size:21px">m = massa</span>\
<span style="font-size:21px">t = tempo</span>\
<span style="font-size:21px">$\rho$ = massa específica</span>




## Unidades de Medida (SI):
### para vazão volumétrica: <span style="font-size:30px">$\frac{m^3}{s}$</span>
### para vazão mássica: <span style="font-size:30px">$\frac{Kg}{s}$</span> 

### Outras uniades:
<span style="font-size:20px">**Para volumétrica:**\
**m³/h** (metros cúbicos por hora)\
**l/min ou l/h** (litros por minuto ou por hora)\
**Nm³/h** (normal metro cúbico por hora, usado para gases em condições normais de temperatura e pressão)\
**SCFH** (_standard cubic feet per hour_ ou pés cubicos por hora padrão, usado em gases)

**Obs: (condições normais ou padrão se refere a 0 °C ou 273,15 K de tempeatura e 1 bar ou 100.000 Pa de Pressão)**

<span style="font-size:20px">**Para mássica:**\
**kg/h** (quilogramas por hora)\
**t/h** (toneladas por hora)\
**lb/h** (libras por hora)</span>

## Anedota Histórica
#### Os relógios de água egípcios, conhecidos como clepsidras, são um dos primeiros instrumentos mecânicos de medição do tempo da humanidade, datando de cerca de 1500 a.C. Eles permitiam medir o tempo à noite, quando os relógios de sol não funcionavam, e usavam o conceito da vazão.

#### Em 3000 a.C., egípcios criaram os "nilômetros", que eram colunas graduadas nas margens do rio para medir a altura da água. Com isso, conseguiam estimar o volume disponível para irrigação. Povos sumérios também monitoravam rios e canais para garantir uma distribuição justa de água.
