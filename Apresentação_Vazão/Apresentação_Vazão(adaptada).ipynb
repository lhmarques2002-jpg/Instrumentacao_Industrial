## Definição Conceitual:

### "A vazão é uma grandeza física utilizada para medir a quantidade de fluido que passa por uma determinada seção de um sistema em um intervalo de tempo."
### Essa quantidade podendo ser descrita tanto pelo volume quanto pela massa do fluido, que por sua vez pode ser tanto um líquido quanto um gaz.
### A vazão volumétrica também pode ser obtida pelo resultado do produto da área seccional pela média da velocidade do fluido.

## Definição Matemática:
### para vazão volumétrica: $\LARGE Q_v = \frac{V}{t}$ ou   $ \LARGE Q_v = S \cdot v$

#### Onde:
$\large Q_v$ <span style="font-size:22px">= Vazão volumétrica</span> \
<span style="font-size:22px">V = volume</span> \
<span style="font-size:22px">t = tempo</span> \
<span style="font-size:22px">S = área</span> \
<span style="font-size:22px">v = velocidade</span> 

### para vazão mássica: $Q_m = \frac{m}{t}$  ou $  Q_m = Q_v \cdot \rho$
#### Onde:
$\large Q_m$ <span style="font-size:22px">= Vazão mássica</span> \
<span style="font-size:22px">m = massa></span> \
<span style="font-size:22px">t = tempo</span> \
$\rho$ <span style="font-size:22px">= massa específica</span> 

## Unidades de Medida (SI):
### para vazão volumétrica: $\LARGE\frac{m^3}{s}$
### para vazão mássica: $\LARGE\frac{Kg}{s}$

### Outras uniades:
<span style="font-size:20px">**Para volumétrica:**\
**m³/h** (metros cúbicos por hora)\
**l/min ou l/h** (litros por minuto ou por hora)\
**Nm³/h** (normal metro cúbico por hora, usado para gases em condições normais de temperatura e pressão)\
**SCFH** (_standard cubic feet per hour_ ou pés cubicos por hora padrão, usado em gases)

**Obs: (condições normais ou padrão se refere a 0 °C ou 273,15 K de tempeatura e 1 bar ou 100.000 Pa de Pressão)**

<span style="font-size:20px">**Para mássica:**\
**kg/h** (quilogramas por hora)\
**t/h** (toneladas por hora)\
**lb/h** (libras por hora)</span>

## Medidores Diretos:

### 1. Medidor de Fluxo por Área Variável

<span style="font-size:22px">São medidores em que a área de escoamento muda automaticamente em função da vazão do fluido. À medida que o fluxo aumenta, um elemento móvel (normalmente um flutuador) desloca-se, aumentando a área de passagem e estabelecendo um novo equilíbrio. A posição desse elemento móvel é então utilizada para indicar a vazão.</span>

### Rotâmetro

#### Princípio de Funcionamento

#### O princípio baseia-se no equilíbrio entre três forças:
<span style="font-size:22px">1. Peso do flutuador\
2. Empuxo exercido pelo fluido\
3. Força de arraste causada pelo escoamento</span>

#### Quando o fluido entra pelo fundo do medidor:

<span style="font-size:22px">O fluido empurra o flutuador para cima e o flutuador sobe até que as forças se equilibrem, quanto maior a vazão, maior será a altura atingida pelo flutuador.
A posição do flutuador é convertida em uma leitura de vazão.</span>

### Vantagens

<span style="font-size:22px">Simplicidade - Não necessita de alimentação elétrica para operação básica.\
Baixa perda de carga - A obstrução ao escoamento é pequena.\
Fácil instalação - Pode ser instalado sem circuitos eletrônicos complexos.\
Baixo custo - Especialmente em comparação com medidores eletromagnéticos ou ultrassônicos.\
Leitura direta - A vazão pode ser observada visualmente.\
Boa confiabilidade - Possui poucas partes móveis.</span>

### Desvantagens
<span style="font-size:22px">Instalação obrigatoriamente vertical - O funcionamento depende da ação da gravidade.\
Sensível à densidade e viscosidade - Mudanças nas propriedades do fluido afetam a calibração.\
Precisão moderada - Normalmente entre ±1% e ±5% do fundo de escala.\
Limitado para altas pressões e temperaturas - Dependendo do material construtivo.\
Dificuldade para fluidos sujos - Partículas podem travar o flutuador.</span>

### Faixas Típicas de Medição

Até centenas de m³/h</span>

### Precisão:

<span style="font-size:22px">Tipicamente de ±1% à ±5%\
dependendo do modelo e da aplicação.</span>

### 2. Medidores Térmicos de Fluxo de Massa

<span style="font-size:22px">Os medidores térmicos de fluxo de massa são instrumentos que medem diretamente a vazão mássica de um fluido utilizando os efeitos da transferência de calor entre um elemento aquecido e o fluido em movimento. São especialmente utilizados para medição de gases, onde oferecem alta precisão e ampla faixa de medição.</span>

#### Princípio de Funcionamento:
#### O funcionamento baseia-se no fenômeno da convecção térmica.
<span style="font-size:22px">Quando um fluido passa por um sensor aquecido
o sensor transfere calor ao fluido.
O fluido remove parte dessa energia térmica e
quanto maior a vazão mássica, maior será a remoção de calor.
A eletrônica do instrumento relaciona essa perda térmica à vazão de massa.</span>

<span style="font-size:22px">O princípio está associado ao conceito de transferência de calor por convecção.</span>

### Métodos de Medição
#### 1. Potência Constante

<span style="font-size:22px">O sensor é mantido aquecido com potência fixa.\
O aumento da vazão provoca maior resfriamento.\
Mede-se a redução da temperatura do sensor.</span>
#### 2. Temperatura Constante

<span style="font-size:22px">O sensor é mantido em uma temperatura fixa acima da temperatura do fluido.\
O circuito aumenta ou reduz a potência de aquecimento.\
A potência necessária para manter a temperatura é proporcional à vazão.</span>

##### Este é o método mais utilizado industrialmente.

### Vantagens:
<span style="font-size:22px">Medição direta de massa - Não requer conversões de vazão volumétrica para mássica.\
Excelente para gases - Apresenta alta precisão em ar, nitrogênio, gás natural, biogás e gases industriais.\
Sem partes móveis - Reduz desgaste mecânico e necessidade de manutenção.\
Baixa perda de carga - Provoca pouca interferência no processo.\
Grande faixa de medição (Turndown) - Relações típicas: 50:1, 100:1, até 1000:1 em modelos avançados\
Boa sensibilidade para baixas vazões - Consegue detectar fluxos muito pequenos com precisão.</span>
### Desvantagens
<span style="font-size:22px">Dependência das propriedades térmicas - Mudanças na composição do fluido afetam a calibração.\
Menos indicado para líquidos - O desempenho é superior em gases.\
Sensibilidade a contaminantes - Sujeira depositada nos sensores altera a transferência de calor.\
Tempo de resposta moderado - Pode ser mais lento que medidores ultrassônicos ou Coriolis.\
Necessidade de calibração específica - Normalmente calibrado para um gás determinado.</span>

### 3. Medidores de Fluxo por Efeito Coriolis

<span style="font-size:22px">Os medidores de fluxo por efeito Coriolis estão entre os instrumentos de medição de vazão mais precisos disponíveis atualmente. Eles medem diretamente a vazão mássica de líquidos, gases e até fluidos multifásicos, utilizando as forças inerciais geradas pelo movimento do fluido em tubos vibratórios.</span>

<span style="font-size:22px">Diferentemente dos medidores volumétricos, que medem volume e posteriormente convertem para massa, o Coriolis mede diretamente a vazão mássica, independentemente de variações de temperatura, pressão ou densidade.</span>

### Conceito Básico

##### O medidor possui um ou mais tubos metálicos que são mantidos vibrando por um excitador eletromagnético.

#### Sem escoamento:

<span style="font-size:22px">Os tubos vibram de forma simétrica.</span>

#### Com escoamento:

<span style="font-size:22px">O fluido que entra e sai dos tubos sofre acelerações e desacelerações gerando forças de Coriolis e essas forças provocam uma torção nos tubos.</span>

##### Quanto maior a vazão mássica maior a torção e maior a diferença de fase detectada pelos sensores.

<span style="font-size:22px">Essa diferença é diretamente proporcional à vazão mássica.</span>

### Equação Fundamental

#### A força de Coriolis pode ser expressa por:

$\LARGE F_c = 2m(v \times w)$

#### Onde:

$\large F_c$<span style="font-size:22px"> = força de Coriolis\
m = massa\
v = velocidade do fluido\
ω = velocidade angular da vibração</span>

<span style="font-size:22px">Na prática, o instrumento mede o deslocamento e a defasagem dos tubos, convertendo-os diretamente em vazão mássica.</span>


### Vantagens
<span style="font-size:22px">Medição direta de massa - Não necessita compensação de pressão ou temperatura.\
Altíssima precisão - Precisões típicas: ±0,05%/±0,1%/±0,2%\
Mede líquidos e gases - Funciona em praticamente qualquer fluido.\
Mede densidade simultaneamente - A frequência natural dos tubos varia conforme a densidade do fluido.\
Sem necessidade de trechos retos - Pode ser instalado próximo a curvas e válvulas.\
Baixa necessidade de manutenção - Não possui partes móveis em contato mecânico.\
Excelente repetibilidade - Uma das melhores entre todos os medidores industriais.</span>

### Desvantagens
<span style="font-size:22px">Alto custo - É um dos medidores mais caros do mercado.\
Peso elevado - Especialmente em grandes diâmetros.\
Sensível a vibrações externas - Pode exigir suportação adequada.\
Limitação em grandes tubulações - Acima de determinados diâmetros o custo torna-se muito elevado.\
Perda de carga moderada - Maior que em medidores ultrassônicos.</span>

### 4. Medidores de Fluxo Mecânicos (Deslocamento Positivo)

<span style="font-size:22px">Os medidores de fluxo por deslocamento positivo (PD – Positive Displacement Flow Meters) são instrumentos que medem a vazão através da divisão do fluido em volumes conhecidos e sucessivos. Cada volume deslocado corresponde a uma quantidade fixa de fluido, permitindo a medição direta do volume total que passou pelo medidor.</span>

<span style="font-size:22px">À medida que esses volumes se enchem e esvaziam, ocorre o deslocamento de componentes mecânicos (engrenagens, pistões, discos, lóbulos etc.), cujo movimento é contado e convertido em vazão ou volume acumulado.</span>

### Princípio de Funcionamento

#### O princípio baseia-se na contagem de volumes conhecidos.

<span style="font-size:22px">1. O fluido entra no medidor.\
2. Uma câmara interna é preenchida.\
3. O fluido desloca um elemento mecânico.\
4. O volume aprisionado é descarregado.\
5. O ciclo se repete continuamente.\
6. O número de ciclos é proporcional ao volume total escoado.</span>

### Vantagens
<span style="font-size:22px">Alta precisão - Precisões típicas entre: ±0,1% e ±0,5%\
Excelente repetibilidade - Muito utilizada para faturamento e transferência comercial.\
Boa medição em baixas vazões - Mantém boa precisão mesmo com escoamentos reduzidos.\
Independência do perfil de velocidade - Não necessita trechos retos longos.\
Funciona bem com fluidos viscosos - Óleos e combustíveis geralmente melhoram o desempenho.\
Medição direta do volume - Não depende de cálculos complexos.</span>

### Desvantagens
<span style="font-size:22px">Possui partes móveis - Maior desgaste mecânico.\
Maior necessidade de manutenção - Comparado a medidores ultrassônicos ou eletromagnéticos.\
Sensível a partículas sólidas - Impurezas podem causar travamentos.\
Perda de carga relativamente alta - Devido às câmaras internas.\

## Medidores Indiretos

### 5. Medidores de Fluxo por Pressão Diferencial

<span style="font-size:22px">Os medidores de fluxo por pressão diferencial (ΔP) são os instrumentos de medição de vazão mais utilizados na indústria devido à sua simplicidade, robustez e baixo custo de instalação. Seu funcionamento baseia-se na relação entre a velocidade do fluido e a queda de pressão provocada por uma restrição inserida na tubulação.</span>

<span style="font-size:22px">Diferentemente dos medidores de deslocamento positivo, Coriolis ou térmicos, estes são considerados medidores indiretos, pois não medem a vazão diretamente. Eles medem uma diferença de pressão (ΔP) e utilizam equações para calcular a vazão correspondente.</span>

### Funcionamento

<span style="font-size:22px">São dispositivos que provocam uma redução controlada da área de escoamento do fluido. Quando o fluido atravessa essa restrição: A velocidade aumenta, a pressão estática diminui e Surge uma diferença de pressão entre montante e jusante.</span>

<span style="font-size:22px">Essa diferença de pressão é medida e convertida em vazão.</span>

### Relação Fundamental

#### A vazão é calculada por: $Q =K\sqrt{ΔP}$
	​
#### Onde:

<span style="font-size:22px">Q = vazão\
K = constante do medidor\
ΔP = pressão diferencial</span>

##### Essa relação mostra que a vazão é proporcional à raiz quadrada da pressão diferencial medida.

### Elementos Secundários

#### Responsável por medir o ΔP.

##### Normalmente utiliza-se um: Transmissor de pressão diferencial (DP Transmitter)
### Elementos de Compensação (quando necessário)

##### Para gases e vapor pode ser necessário medir: Pressão absoluta e Temperatura

##### A fim de compensar variações de densidade.

#### O tubo de Pitot mede a velocidade local do fluido a partir da diferença entre pressão total e pressão estática.

### Equação da Velocidade

#### A velocidade é determinada por: $v=\sqrt{\frac{2ΔP}{ρ}}$
#### Onde:

<span style="font-size:22px">v = velocidade\
ΔP = pressão diferencial\
ρ = densidade do fluido</span>

### Vantagens
<span style="font-size:22px">Baixíssima perda de carga.\

### Desvantagens

### Precisão

#### Tipicamente: ±1% à ±5%

### 6.Medidores de Fluxo por Eletromagnetismo

<span style="font-size:22px">Os medidores de fluxo eletromagnéticos, também chamados de medidores magnéticos ou magmeters, são instrumentos que medem a vazão de fluidos eletricamente condutivos utilizando a indução eletromagnética..</span>

<span style="font-size:22px">Diferentemente dos medidores por pressão diferencial, eles realizam uma medição direta da velocidade média do fluido, sem criar obstruções significativas ao escoamento.</span>

### Princípio de Funcionamento

#### Quando um fluido condutivo atravessa um campo magnético:

<span style="font-size:22px">1. O campo magnético é gerado por bobinas instaladas no corpo do medidor.\
2. O fluido em movimento corta as linhas do campo magnético.\
3. Surge uma tensão elétrica entre dois eletrodos.\
4. Essa tensão é proporcional à velocidade do fluido.\
5. A velocidade é convertida em vazão.</span>

### Lei de Faraday

#### A tensão induzida pode ser expressa por: $E=B⋅L⋅v$

#### Onde:

<span style="font-size:22px">E = tensão induzida (V)\
B = intensidade do campo magnético (T)\
L = distância entre os eletrodos (m)\
v = velocidade média do fluido (m/s)</span>

#### Como B e L são constantes do instrumento:

### $E∝v$

#### Portanto, a tensão medida é diretamente proporcional à velocidade do fluido.

### Vantagens
<span style="font-size:22px">Sem partes móveis - Reduz desgaste e manutenção.\
Sem obstruções internas - Praticamente não gera perda de carga adicional.\
Alta precisão - Precisões típicas: ±0,2% à ±0,5%\
Excelente para fluidos sujos - Pode medir: Lodo, Polpas minerais, Efluentes\
Grande faixa de medição - Turndown frequentemente superior a 100:1.\
Independente da densidade - Mudanças de densidade não afetam significativamente a medição.\
Independente da viscosidade - Excelente desempenho com fluidos viscosos.</span>
### Desvantagens
<span style="font-size:22px">Requer fluido condutivo - Principal limitação da tecnologia.\
Não mede gases - Necessita contato elétrico com o fluido.\
Não mede hidrocarbonetos puros - Combustíveis geralmente possuem baixa condutividade.\
Custo elevado - Maior que placas de orifício e rotâmetros.\
Requer alimentação elétrica - Não funciona de forma puramente mecânica.</span>

### Precisão: ±0,2% a ±0,5%

Esgoto\

### 7.Medidores de Fluxo Ultrassônicos

<span style="font-size:22px">Os medidores de fluxo ultrassônicos utilizam ondas sonoras de alta frequência (ultrassom) para determinar a velocidade e a vazão de um fluido.</span>

### Princípio de Funcionamento

#### Os medidores ultrassônicos operam analisando o comportamento das ondas sonoras propagadas através do fluido.

#### Existem dois métodos principais:

<span style="font-size:22px">Tempo de Trânsito (Transit Time)\
Efeito Doppler</span>

### 7.1. Medidor Ultrassônico por Tempo de Trânsito

#### É o método mais utilizado em líquidos limpos.

### Funcionamento

#### Dois transdutores são instalados em lados opostos da tubulação.

#### Um sinal é enviado:

<span style="font-size:22px">No sentido do fluxo.\
Contra o sentido do fluxo.</span>

#### A velocidade do som é afetada pelo movimento do fluido.

#### Consequentemente:

<span style="font-size:22px">O tempo de propagação a favor do fluxo diminui.\
O tempo de propagação contra o fluxo aumenta.</span>

#### A diferença entre esses tempos permite calcular a velocidade do fluido.

### Vantagens
<span style="font-size:22px">Alta precisão.\
Sem perda de carga.\
Excelente para líquidos limpos.\
Não possui partes móveis.</span>
### Desvantagens
<span style="font-size:22px">Sensível à presença de bolhas.\
Exige fluido relativamente homogêneo.\
Custo elevado.</span>

### 7.2. Medidor Ultrassônico por Efeito Doppler

#### Utilizado quando o fluido contém partículas sólidas ou bolhas em suspensão.

### Funcionamento

#### O transdutor emite uma onda ultrassônica. Essa onda é refletida pelas partículas ou bolhas presentes no fluido e o sinal refletido retorna com uma frequência diferente devido ao efeito Doppler.

#### A variação de frequência é utilizada para calcular a velocidade do escoamento.

### Equação Conceitual

#### A frequência Doppler pode ser expressa por:

#### Δf∝v

#### Onde:

<span style="font-size:22px">Δf = variação de frequência\
v = velocidade do fluido</span>

### Vantagens
<span style="font-size:22px">Funciona com líquidos sujos.\
Adequado para lodos e efluentes.\
Instalação simples.</span>
### Desvantagens
<span style="font-size:22px">Menor precisão.\
Requer partículas ou bolhas para funcionar.</span>

### Tipos de Instalação
### Clamp-On

#### Os sensores são fixados externamente à tubulação.

### Vantagens
<span style="font-size:22px">Não exige corte da tubulação.\
Instalação rápida.\
Sem contato com o fluido.</span>
### Aplicações
<span style="font-size:22px">Sistemas temporários.\
Auditorias energéticas.\
Testes de campo.</span>

### In-Line

#### Os transdutores ficam em contato direto com o fluido.

### Vantagens
<span style="font-size:22px">Maior precisão.\
Menor influência externa.</span>
### Aplicações
<span style="font-size:22px">Processos industriais permanentes.</span>

### Sensores Necessários

#### Os medidores ultrassônicos normalmente utilizam:

#### Transdutores Ultrassônicos - Emissão e recepção das ondas sonoras.

### Desvantagens
<span style="font-size:22px">Custo elevado - Superior ao de placas de orifício e rotâmetros.\

Exige perfil de escoamento adequado - Normalmente necessita trechos retos.\

### 8.Medidores de Fluxo por Força de Arrasto

<span style="font-size:22px">Os medidores de fluxo por força de arrasto (Drag Force Flow Meters) medem a vazão de um fluido a partir da força exercida pelo escoamento sobre um corpo fixo inserido na tubulação. Essa força, denominada força de arrasto, é proporcional à velocidade do fluido e pode ser convertida em vazão volumétrica ou mássica.</span>

### Princípio de Funcionamento

#### O funcionamento baseia-se na força de arrasto gerada pelo escoamento sobre um obstáculo.

#### Quanto maior a velocidade do fluido:

<span style="font-size:22px">Maior a força exercida.\
Maior a deformação do sensor.\
Maior o sinal gerado.</span>

### Força de Arrasto

#### A força de arrasto pode ser expressa por:

$\large F_d = \frac{1}{2}C_dρAv^2$

#### Onde:

$\large F_d$<span style="font-size:22px"> = força de arrasto (N)</span>\
$\large C_d$<span style="font-size:22px"> = coeficiente de arrasto</span>\
<span style="font-size:22px">ρ = densidade do fluido (kg/m³)\
A = área projetada do sensor (m²)\
v = velocidade do fluido (m/s)</span>

#### Observa-se que a força é proporcional ao quadrado da velocidade.

### Vantagens
<span style="font-size:22px">Construção simples - Possui poucos componentes mecânicos.\
Aplicável a líquidos, gases e vapor - Grande versatilidade de aplicação.\
Boa resistência mecânica - Adequado para ambientes industriais severos.\
Instalação em grandes diâmetros - Particularmente vantajoso em tubulações extensas.\
Baixa manutenção - Não possui partes móveis rotativas.\
Boa faixa de medição - Turndown típico: 10:1, 20:1</span>
### Desvantagens
<span style="font-size:22px">Necessita conhecimento da densidade - Como a força depende da densidade do fluido.\
Sensível ao perfil de escoamento - Curvas, válvulas e turbulências podem afetar a precisão.\
Introduz obstrução ao fluxo - O elemento sensor permanece dentro da tubulação.\
Influência de vibrações mecânicas - Tubulações vibrando podem gerar erros.</span>

### Precisão
#### Na faixa de: ±1% à ±3%

### Aplicações
#### Vapor Industrial

<span style="font-size:22px">Caldeiras\
Redes de distribuição\
Processos térmicos</span>

<span style="font-size:22px">Linhas industriais\
Sistemas pneumáticos</span>

<span style="font-size:22px">Nitrogênio\

### 9. Medidores de Fluxo por Vórtice

<span style="font-size:22px">Os medidores de fluxo por vórtice (Vortex Flow Meters) são instrumentos que medem a vazão a partir da formação periódica de vórtices gerados quando um fluido escoa ao redor de um obstáculo inserido na tubulação.</span>


### Princípio de Funcionamento

#### O medidor possui um obstáculo fixo instalado no interior da tubulação.

#### À medida que o fluido passa:

<span style="font-size:22px">1. O fluxo encontra o obstáculo.\
2. Formam-se vórtices alternadamente à esquerda e à direita.\
3. Surge uma oscilação periódica de pressão.\
4. Um sensor detecta essa oscilação.\
5. A frequência medida é utilizada para determinar a velocidade do escoamento.</span>

### Rua de Von Kármán

#### O fenômeno foi estudado por Theodore von Kármán e descreve a formação alternada de vórtices atrás de um obstáculo submetido a um fluxo.

#### A frequência dos vórtices é relacionada à velocidade do fluido pelo Número de Strouhal.

### Equação Fundamental

#### A relação básica é:

$\large f=St⋅\frac{v}{d}$
	​


#### Onde:

<span style="font-size:22px">f = frequência dos vórtices (Hz)\
St = número de Strouhal\
v = velocidade do fluido (m/s)\
d = largura do obstáculo (m)</span>

#### Como St e d são constantes do medidor:

#### f∝v

#### Logo, a frequência é diretamente proporcional à velocidade do fluido.

### Sensores Adicionais

#### Sensor de Temperatura - Determina a temperatura do fluido.

#### Sensor de Pressão - Permite calcular a densidade.

### Vantagens
<span style="font-size:22px">Sem partes móveis - Reduz manutenção e desgaste.\
Boa precisão - Precisões típicas: ±0,5% à ±1%\
Funciona com líquidos, gases e vapor - Grande versatilidade.\
Excelente para vapor - Uma das tecnologias mais utilizadas para medição de vapor industrial.\
Boa resistência a altas temperaturas - Alguns modelos operam acima de 400 °C.\
Baixa manutenção - Longa vida útil.</span>
### Desvantagens
<span style="font-size:22px">Sensível a vibrações externas - Vibrações da tubulação podem interferir na medição.\
Necessita velocidade mínima - Abaixo de determinada velocidade não ocorre formação estável de vórtices.\
Não indicado para fluidos muito viscosos - A viscosidade elevada prejudica a formação dos vórtices.\
Exige trechos retos - Normalmente requer: 10 a 30 diâmetros a montante e 5 a 10 diâmetros a jusante\
Pode gerar pequena perda de carga - Devido ao obstáculo inserido.</span>

### Precisão: ±0,5% a ±1%
